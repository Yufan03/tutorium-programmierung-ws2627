### Dictionary

```python
# die Weisen wie ein dictionary erzeugt wird

a = {"a":1, "b":2} # einfachste erzeugung
print(a)

a = dict([("a",1),("b",2)]) # dict() + eine liste von tupeln aus key-value paaren
print(a)
```

```python
# auf den Wert durch Schlüssel zugreifen
a["b"]
```

```python
# dictionary ist veränderbar
a["b"] = 0
a
```

```python
# einen neuen wert hinzufügen
a["c"] = 3
a
```

```python
# einen bestimmten wert entfernen
del a["c"]
a
```

```python
# der Schlüssel muss unveränderbar (tupel, string usw.) sein, sonst Fehler ausgelöst
a[[1,2]] = 3
```

```python
# auf einen schlüssel zugreifen der nicht existiert
a["d"]
```

```python
# den fehler durch die methode get vermeiden
a.get("d", 0) # 0 ist der default ausgegebene Wert falls der schlüssel nicht vorhand ist 
```

```python
# key, values, key-value paare checken
print(a.keys())
print(a.values())
print(a.items())
# die suffix sind alle methode aus einer klasse (hier dictionary), mehr dazu später
```

```python
# eine der wichtigsten anwendungen des dictionarys - häufigkeiten berechnen

beispiel_satz = "ich glaub, dass er glaubt, dass sie glaubt, dass es nicht wahr ist"


# hier ist die vorverarbeitung des satzes, damit die vorkommnisse gezählt werden können
import re 
zerlegter_satz = re.split(r'\W+', beispiel_satz)
print(zerlegter_satz)

zähler = {}
for wort in zerlegter_satz:
    if wort not in zähler: # um den keyerorr zu vermeiden haben wir if bedingung in die schleife eingebracht
        zähler[wort] = 1
    else:
        zähler[wort] += 1
print(zähler)
```

```python
# durch die funktion defaultdict ist die for-schleife noch zu vereinfachen also ohne if bedingung 

# ABER! diese funktion ist in der standardbibliothek, vor der anwendung müssen wir's importieren
from collections import defaultdict # das ist spezifischer import
a = defaultdict(int) # das argument verlangt einen datentyp

"""
anders als spezifischer import können wir auch gesamtes modul importieren, 
dafür rufen wir die funktion umständlich auf mit dem namen collections.defaultdict()
"""

# import collections
# a = collections.defaultdict(int)


for wort in zerlegter_satz:
    a[wort] += 1
print(a)
```

### Dateiverarbeitung

```python
# die datei, die demnächst geöffnet werden soll, muss ja vorher existieren. und zwar im selben arbeitsverzeichnis
# deshalb erstellen wir zuerst vorab ne .txt datei 

# der text ist leer, lass uns erstmal schreiben
a = open("beispiel.txt", 'w') #'w' steht für write --> überschreiben
a.write("hallo world\nciao")
a.close()
```

```python
a = open("beispiel.txt", 'a') # 'a' steht für append
a.write("\nhi, ich bin wieder da")
a.close()
```

```python
# wir können jetzt schauen was reingeschrieben wird
a = open("beispiel.txt", 'r') # 'r' steht für read
behaelter = a.read() # 'read' lädt den ganzen text
a.close()
behaelter
```

```python
a = open("beispiel.txt", 'r')
behaelter1 = a.readline() # 'readline' lädt nur eine zeile bzw die nächste zeile
a.close()
behaelter1
```

```python
a = open("beispiel.txt", 'r')
behaelter2 = a.readlines() # 'readlines'nimmt alle zeilen in eine liste auf
a.close()
behaelter2
```

```python
# es ist viel zu umständlich, die datei immer zu öffnen und schließen
with open("beispiel.txt", 'r') as file:
    behaelter3 = file.readline()
    behaelter4 = file.readline()
    # nach all diesen bearbeitungen schließt es sich automatisch
print(behaelter3)
print(behaelter4)
```

```python
# die zeilen in einer textdatei sind auch zu iterieren
with open("beispiel.txt", 'r') as file:
    for zeile in file:
        print(zeile)
```

```python
# die Argumente sind nicht der fesetn reihenfolge nach angegeben --> fehler
with open("beispiel.txt", 'r', 'utf-8') as file: # 'utf-8' wird buffering zugewiesen, buffering string verlangt
    pass
```

### Klasse

jedes python-objekt ist eine instanz einer klasse, z.B. objekt 'list' ist von der Klasse 'List' 

```python
class Bankkonto: # der erste Buchstabe des namen muss großgeschrieben sein
    def __init__(self, besitzer, saldo): # self ist das objekt selbst
        self.besitzer = besitzer
        self.saldo = saldo
        # alle variablen in der initialisiereung sind attribute von dem objekt 
    def saldo_anzeigen(self):
        print(f'kontostand: {self.saldo}')
```

```python
# die Erzeugung eines Objekts der Klasse
konto = Bankkonto("Ken", 500) # was ist self 
```

```python
# das attribut (bzw. variable) ohne klammern aufrfufen
konto.besitzer
```

```python
konto.saldo
```

```python
# die Methode (bzw. funktion) aufrufen
konto.saldo_anzeigen()
```

##### Fernverbindung:
Mit ssh wird eine Fernverbindung zu einem anderen Rechner hergestellt. Sobald die Verbindung steht, tippen wir Befehle in unser lokales Terminal ein, aber sie werden direkt auf dem entfernten Server ausgeführt.

##### Einsatz von * und ? für Platzhalter
1. der Stern * steht für beliebig viele Zeichen oder kein Zeichen, z.B. ls *.py
2. das Fragezeichen ? steht für exakt ein einzelnes Zeichen, z.B. ls woche?.txt

##### Zugriffsrechte
Auf einem Server, auf dem viele Mitarbeiter gleichzeitig arbeiten, schützt diese Trennung die Daten voneinander:
1. Besitzer (u): er hat Vollzugriff (rwx) auf die Abgaben also z.B. einer Datei
2. Gruppe (g): Mitglieder derselben Übungsgruppe erhalten Leserechte (r--) oder Schreibrechte (rw-) auf gemeinsame Projektordner
3. Others (o): die outsiders auf demselben Server erhalten keine Rechte (---) und können die Dateien weder lesen noch verändern

##### die Rechtsymbole
1. r --> read
2. w --> write
3. x --> execute

##### die Bedeutung eines 10-stelligen strings in der ausgabe von ls -l

beispiel_string: -rwxr-xr--

1. Erstes Zeichen (- oder d): Typ (- = normale Datei, d = Directory/Ordner)
2. Zeichen 2–4 (rwx): Rechte des Besitzers 
3. Zeichen 5–7 (r-x): Rechte der Gruppe 
4. Zeichen 8–10 (r--): Rechte für alle anderen
