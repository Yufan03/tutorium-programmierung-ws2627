### Übung2

Klassendefinition: Erstelle eine Klasse namens Mensakarte.   

Initialisierung: Schreibe die __init__-Methode, die das Objekt (self) sowie die Parameter matrikelnummer und guthaben entgegennimmt. Speichere diese beiden Werte als Attribute der Instanz (z. B. self.guthaben).   

Guthaben aufladen: Erstelle eine Methode aufladen(self, betrag). Diese Methode soll aufgerufen werden können, um den übergebenen betrag zum aktuellen self.guthaben zu addieren.

Essen bezahlen: Erstelle eine Methode bezahlen(self, preis). Diese soll prüfen (mit einem simplen if/else), ob das Guthaben für den übergebenen preis ausreicht. Wenn ja, wird der Preis vom Guthaben abgezogen. Wenn nein, soll eine Meldung per print ausgegeben werden, dass das Guthaben nicht reicht.

Status ausgeben: Schreibe eine Methode status_anzeigen(self), die analog zur Methode saldo_anzeigen aus dem Skript funktioniert und die Matrikelnummer sowie das aktuelle Guthaben als formatierten f-String auf der Konsole ausgibt.   

Anwendung: Erzeuge ein Objekt der Klasse Mensakarte für einen fiktiven Studenten (z. B. mit 10 Euro Startguthaben). Teste die Klasse, indem du die Karte auflädst, ein Gericht bezahlst und am Ende die Methode status_anzeigen() aufrufst, um das Restguthaben zu überprüfen.   

```python
class Mensakarte:
    def __init__(self, matrikelnummer, guthaben):
        self.matrikelnummer = matrikelnummer
        self.guthaben = guthaben
    def aufladen(self, betrag):
        self.guthaben += betrag
    def bezahlen(self, preis):
        if self.guthaben < preis:
            print('das guthaben reicht nicht mehr')
        self.guthaben -= preis
    def status_anzeigen(self):
        print(f'das aktuelle guthaben: {self.guthaben}\ndie Martrikelnummer lautet {self.matrikelnummer}')
        
        
```

```python
ken = Mensakarte('12345',10)
```

```python
ken.aufladen(100)
```

```python
ken.guthaben
```

```python
ken.bezahlen(6)
```

```python
ken.status_anzeigen()
```
